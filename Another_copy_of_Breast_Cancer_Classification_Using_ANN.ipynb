```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import sklearn.datasets
```

**SPLITTING THE DATASET INTO FEATURES AND TARGET**

```python
# x: This is a variabel that contains all the independent features
# y: This is a variabel that contains only the target
x=cancer_df.drop(columns=['label'], axis=1)
y=cancer_df['label']
```

```python
#priting out the values in the x variable
print(x)
```

```python
# printing out the values in the y variable
print(y)
```

**DATA SPLICING**

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.1, random_state=2)
```

**DATA SCALING**

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
xtrain_scaled = scaler.fit_transform(xtrain)
xtest_scaled = scaler.transform(xtest)
```

```python
# viewing the scaled data for xtrain and xtest
xtrain_scaled
```

```python
xtest_scaled
```

**IMPORT DEEP LEARNING LIBRARIES**

nn-breast-cancer.png

```python
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

**BUILDING THE ARCHITECTURE OF THE NEURAL NETWORK**

```python
model = keras.Sequential(
    [
        keras.layers.Flatten(input_shape=(30,)),  # This is the input layer
        keras.layers.Dense(15, activation="relu"), # This is the hidden layer with 20 neurons
        keras.layers.Dense(15, activation="relu"), # This is the hidden layer with 20 neurons
        keras.layers.Dense(2, activation="softmax") # This is the output layer for two classes classification
    ]
)
```

```python
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
```

**TRAINING THE NEURAL NETWORK**

```python
history =model.fit(xtrain_scaled, ytrain, validation_split=0.1, epochs=15)
```

**EVALUATING MODEL TEST ACCUARCY**

```python
test_loss, test_acc = model.evaluate(xtest_scaled, ytest)
rounded_test_acc = round(test_acc, 2)
print(f"The model Test Accuracy:, {rounded_test_acc}")
```

```python

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()

plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()
```

**BUILDING THE PREDICTIVE SYSTEM**

```python
input_data = (12.05,14.63,78.04,449.3,0.1031,0.09092,0.06592,0.02749,0.1675,0.06043,0.2636,0.7294,1.848,19.87,0.005488,0.01427,0.02322,0.00566,0.01428,0.002422,13.76,20.7,89.88,582.6,0.1494,0.2156,0.305,0.06548,0.2747,0.08301)

# change the input data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the numpy array as we are predicting for one datapoint
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

# standardizing the input data
input_data_std = scaler.transform(input_data_reshaped)

prediction = model.predict(input_data_std)
print(prediction)

prediction_label = [np.argmax(prediction)]
print(prediction_label)

if(prediction_label[0] == 0):
  print('The tumor is Benign')

else:
  print('The tumor is Malignant')
```
